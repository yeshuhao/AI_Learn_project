```python
import torch

data = torch.tensor([[1,2],[3,4]])     #tensor为张量
data
```

```python
#使用随机张量或常量值
# shape 是张量维度的元组
shape = (2,3,)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor} \n")
```

```python
#基于现有tensor构建，但使用新值填充
m = torch.ones(5,3, dtype=torch.double)
n = torch.rand_like(m, dtype=torch.float)

#获取tensor的大小
print(m.size())    #torch.Size([5,3])

#均匀分布
print(torch.rand(5,3))
#标准正态分布
print(torch.randn(5,3))
#离散正态分布
print(torch.normal(mean=.0,std=1.0,size=(5,3)))
#线性间隔向量（返回一个1维向量，包含在区间start和end上均匀间隔约steps个点）
print(torch.linspace(start=1,end=10,steps=20))
```

```python
tensor = torch.rand(3,4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

```python
tensor = torch.ones(4,4)
print('First row:', tensor[0])
print('First column:', tensor[:,0])
print('Last colum:', tensor[...,-1])     #[...,-1]表示取所有行中的最后一列
tensor[:,1] = 0
print(tensor)
```

```python
#张量的拼接
t1 = torch.cat([tensor,tensor,tensor], dim=1)
print(t1)
print(t1.shape)
```

```python
import torch
tensor = torch.arange(1,10,dtype=torch.float32).reshape(3,3)

#计算两个张量之间矩阵乘法的几种方式， y1, y2, y3 最后的值是一样的 dot
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)
print(y1)
print(y2)

y3 = torch.rand_like(tensor)
torch.matmul(tensor,tensor.T,out=y3)
print(y3)

#PyTorch中也完全支持点积运算
```

```python
#转换为python类型（求单一结构值）
agg = tensor.sum()
agg_item = agg.item()
print(agg_item, type(agg_item))
```

```python
#In-place操作（直接修改张量里面的内容）
tensor.add_(5)    #模型计算时，该方法不太推荐，因为会丢失历史记录
# tensor = tensor + 5
# tensor += 5
print(tensor)
```
